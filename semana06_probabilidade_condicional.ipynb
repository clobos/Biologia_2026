```python
import pandas as pd

# ---------------------------------------------------------
# 1. Dados: tabela 2 x 2
# Linhas: condição real
# Colunas: resultado do teste
# ---------------------------------------------------------

tabela = pd.DataFrame(
    {
        "Teste positivo": [24, 5],
        "Teste negativo": [6, 65]
    },
    index=["Com COVID-19", "Sem COVID-19"]
)

# Totais marginais
tabela["Total"] = tabela.sum(axis=1)
tabela.loc["Total"] = tabela.sum(axis=0)

print("Tabela de contingência (frequências):")
display(tabela)

```

```python
# ---------------------------------------------------------
# 2. Probabilidades conjuntas e marginais
# ---------------------------------------------------------

n = tabela.loc["Total", "Total"]

probabilidades = tabela / n

print("Tabela de probabilidades conjuntas e marginais:")
display(probabilidades)

```

```python
# Frequências das quatro células
VP = tabela.loc["Com COVID-19", "Teste positivo"]
FN = tabela.loc["Com COVID-19", "Teste negativo"]
FP = tabela.loc["Sem COVID-19", "Teste positivo"]
VN = tabela.loc["Sem COVID-19", "Teste negativo"]

# ---------------------------------------------------------
# 3. Eventos
# C  = pessoa com COVID-19
# Cc = pessoa sem COVID-19
# Tp = teste positivo
# Tn = teste negativo
# ---------------------------------------------------------

P_C = (VP + FN) / n
P_Cc = (FP + VN) / n

P_Tp = (VP + FP) / n
P_Tn = (FN + VN) / n

# Probabilidades conjuntas = interseções
P_C_inter_Tp = VP / n
P_C_inter_Tn = FN / n
P_Cc_inter_Tp = FP / n
P_Cc_inter_Tn = VN / n

# ---------------------------------------------------------
# 4. Probabilidades condicionais
# ---------------------------------------------------------

# Condição real dado o resultado do teste
P_C_dado_Tp = VP / (VP + FP)
P_Cc_dado_Tp = FP / (VP + FP)

P_C_dado_Tn = FN / (FN + VN)
P_Cc_dado_Tn = VN / (FN + VN)

# Resultado do teste dado a condição real
P_Tp_dado_C = VP / (VP + FN)
P_Tn_dado_C = FN / (VP + FN)

P_Tp_dado_Cc = FP / (FP + VN)
P_Tn_dado_Cc = VN / (FP + VN)

# ---------------------------------------------------------
# 5. União de eventos
# P(C ∪ T+) = P(C) + P(T+) - P(C ∩ T+)
# ---------------------------------------------------------

P_C_uniao_Tp = P_C + P_Tp - P_C_inter_Tp
P_C_uniao_Tn = P_C + P_Tn - P_C_inter_Tn

# ---------------------------------------------------------
# 6. Probabilidades complementares
# ---------------------------------------------------------

P_complemento_C = 1 - P_C
P_complemento_Tp = 1 - P_Tp
P_complemento_C_inter_Tp = 1 - P_C_inter_Tp

# ---------------------------------------------------------
# 7. Indicadores de desempenho diagnóstico
# ---------------------------------------------------------

sensibilidade = P_Tp_dado_C
especificidade = P_Tn_dado_Cc
valor_preditivo_positivo = P_C_dado_Tp
valor_preditivo_negativo = P_Cc_dado_Tn
acuracia = (VP + VN) / n

# ---------------------------------------------------------
# 8. Resultados
# ---------------------------------------------------------

resultados = {
    "P(C) = P(com COVID-19)": P_C,
    "P(Cᶜ) = P(sem COVID-19)": P_Cc,
    "P(T+) = P(teste positivo)": P_Tp,
    "P(T-) = P(teste negativo)": P_Tn,
    "P(C ∩ T+) = verdadeiro positivo": P_C_inter_Tp,
    "P(C ∩ T-) = falso negativo": P_C_inter_Tn,
    "P(Cᶜ ∩ T+) = falso positivo": P_Cc_inter_Tp,
    "P(Cᶜ ∩ T-) = verdadeiro negativo": P_Cc_inter_Tn,
    "P(C | T+) = valor preditivo positivo": P_C_dado_Tp,
    "P(Cᶜ | T+)": P_Cc_dado_Tp,
    "P(C | T-)": P_C_dado_Tn,
    "P(Cᶜ | T-) = valor preditivo negativo": P_Cc_dado_Tn,
    "P(T+ | C) = sensibilidade": P_Tp_dado_C,
    "P(T- | C) = taxa de falso negativo": P_Tn_dado_C,
    "P(T+ | Cᶜ) = taxa de falso positivo": P_Tp_dado_Cc,
    "P(T- | Cᶜ) = especificidade": P_Tn_dado_Cc,
    "P(C ∪ T+)": P_C_uniao_Tp,
    "P(C ∪ T-)": P_C_uniao_Tn,
    "P(Cᶜ)": P_complemento_C,
    "P(T-) = P((T+)ᶜ)": P_complemento_Tp,
    "P((C ∩ T+)ᶜ)": P_complemento_C_inter_Tp,
    "Acurácia": acuracia
}

resultado_df = pd.DataFrame(
    resultados.items(),
    columns=["Probabilidade ou medida", "Valor"]
)

resultado_df["Percentual"] = resultado_df["Valor"] * 100

display(resultado_df.style.format({
    "Valor": "{:.4f}",
    "Percentual": "{:.2f}%"
}))
```
