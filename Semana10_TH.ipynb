# Teste para 1 média

```python
import pandas as pd
url = ("https://raw.githubusercontent.com/"
"allisonhorst/palmerpenguins/main/"
"inst/extdata/penguins.csv")
df = pd.read_csv(url)
# Filtrar espécie e ilha
ade = df[(df.species == "Adelie") &
(df.island == "Torgersen")]
# Remover NA apenas na variável de interesse
massa = ade["body_mass_g"].dropna()
n = len(massa)
print(f"n = {n}")
```

```python
import numpy as np
from scipy import stats
mu0 = 3700
x_bar = massa.mean()
s    = massa.std(ddof=1)
n    = len(massa)
# Estatística t
t_obs = (x_bar - mu0) / (s / np.sqrt(n))
print(f"X_barra= {x_bar:.1f} g")
print(f"S = {s:.1f} g, n = {n}")
print(f"t_obs = {t_obs:.4f}")
```

```python
res = stats.ttest_1samp(
massa, popmean=mu0
)
print(res)
```

```python
from scipy import stats

# Graus de liberdade
df_degrees = 50

# Calculando os percentis (usando probabilidades 0.025 e 0.975)
percentil_2_5 = stats.t.ppf(0.025, df=df_degrees)
percentil_97_5 = stats.t.ppf(0.975, df=df_degrees)

print(f"Percentil 2.5: {percentil_2_5:.4f}")
print(f"Percentil 97.5: {percentil_97_5:.4f}")
```

 # Teste para 1 variância

```python
import numpy as np
from scipy import stats
sigma0_sq = 500**2   # 250000 g²
s_sq = massa.var(ddof=1)
n    = len(massa)
# Estatística qui-quadrado
q_obs = (n - 1) * s_sq / sigma0_sq
# Pontos críticos (bilateral, α = 0.05)
alpha = 0.05
q_lower = stats.chi2.ppf(alpha/2,   df=n-1)
q_upper = stats.chi2.ppf(1-alpha/2, df=n-1)
print(f"S² = {s_sq:.1f} g²")
print(f"Q  = {q_obs:.4f}")
print(f"q_lower={q_lower:.4f}, q_upper={q_upper:.4f}")
```

# Teste para 1 proporção

```python
import numpy as np
from scipy import stats
# Filtrar sexo registrado
ade = df[(df.species == "Adelie") &
(df.island == "Torgersen")]
ade_sexo = ade.dropna(subset=["sex"])
n = len(ade_sexo)
k = (ade_sexo["sex"] == "female").sum()
p0 = 0.50
p_hat = k / n
print(f"k = {k}, n = {n}")
print(f"p_hat = {p_hat:.4f}")
z_obs = (p_hat - p0) / np.sqrt(
p0*(1-p0)/n)
print(f"z_obs = {z_obs:.4f}")
```
